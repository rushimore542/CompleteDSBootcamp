## logical Operator

```python
## AND,Not,OR
## AND:For AND operator both condition should be true.Then only you get answer true
x=True
y=True
a=True
b=False
result1=x and y
result2=a and b
print(result1,result2)
```

```python
## OR :Any one condition between two conditions is giving the boolean output true then you will get output as true.
## If both input conditions are giving boolean output false then only the output of OR operation is False.
x=True
y=False
r1=x or y

a=False
b=True
r2=a or b

m=False
n=False
r3=m or n
print(r1,r2,r3) 
```

```python
## Not operator:It is basically reverse of the boolean operator
x=False
y=not x
print(y)
```
